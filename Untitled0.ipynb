```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("sakshisatre/ice-cream-sales-dataset")

print("Path to dataset files:", path)
```

```python
import os
import pandas as pd

# Listar los archivos en la ruta de descarga
archivos = os.listdir(path)
print("Archivos encontrados:", archivos)

# Buscar el archivo CSV principal
csv_files = [f for f in archivos if f.endswith('.csv')]
if csv_files:
    csv_path = os.path.join(path, csv_files[0])
    df = pd.read_csv(csv_path)
    print(f"\nPrimeras 5 filas de: {csv_files[0]}")
    display(df.head())
else:
    print("No se encontró ningún archivo CSV en la ruta.")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configurar o estilo do gráfico
sns.set_theme(style="whitegrid")

# Criar a figura
plt.figure(figsize=(10, 6))

# Gerar o gráfico de regressão linear
sns.regplot(data=df, x='Temperature', y='Revenue', scatter_kws={'alpha':0.6}, line_kws={'color':'red'})

# Títulos e rótulos
plt.title('Regressão Linear: Temperatura vs Faturamento (Revenue)', fontsize=14)
plt.xlabel('Temperatura (°C)', fontsize=12)
plt.ylabel('Faturamento (Revenue)', fontsize=12)

# Mostrar o gráfico
plt.show()
```
